```python
# 12-24 25-44 45-54 55+
def f(x):
    if x > 54:
        return 5
    if x > 44:
        return 4
    if x > 24:
        return 3
    if x > 11:
        return 2
    return 1
```

```python
demo['age_group'] = demo['age'].apply(lambda x: f(x))
```
